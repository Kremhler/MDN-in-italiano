# Impara lo sviluppo web 

Benvenuto in **MDN Learning Web Development** (conosciuto anche come Learn). Questa risorsa offre un insieme strutturato di tutorial che insegnano le competenze e le pratiche essenziali per diventare un front end developer di successo, insieme a sfide ed ulteriori risorse consigliate.

## Informazioni 


-	Insegna le competenze e le conoscenze essenziali di cui ogni front end developer ha bisogno per avere successo nella carriera e rimanere rilevante nel settore, come definito nel [MDN Curriculum](https://developer.mozilla.org/en-US/curriculum/).
-	Creato dalla comunità MDN e perfezionato grazie ai contributi di studenti, insegnanti e sviluppatori della più ampia comunità web.
-	Progettato per portarti da “principiante” a “a tuo agio” (non da “principiante” a “esperto”), dandoti abbastanza conoscenze per utilizzare risorse più avanzate (come il [resto di MDN](https://developer.mozilla.org/en-US/)).


## Non sai da dove iniziare?
##### Mai programmato prima?
I nostri moduli [Getting started](https://developer.mozilla.org/en-US/docs/Learn_web_development/Getting_started) forniscono tutorial di configurazione, concetti essenziali e informazioni di base per chi parte da zero. Se sei un principiante assoluto (cioè non hai ancora installato un editor di codice né scritto una riga di codice), dovresti iniziare da qui.
##### Vuoi padroneggiare le basi?
I nostri [Core modules](https://developer.mozilla.org/en-US/docs/Learn_web_development/Core) offrono un insieme strutturato di tutorial che insegnano le competenze e le pratiche fondamentali per diventare un front end developer di successo.
##### Oltre le basi?
I nostri [Extension modules](https://developer.mozilla.org/en-US/docs/Learn_web_development/Extensions) coprono competenze aggiuntive utili mentre inizi ad ampliare le tue conoscenze e sviluppare specializzazioni. Passa a questi dopo aver completato i Core.
##### Lavori in una scuola?
Usa i nostri moduli per guidare la tua didattica; consulta anche la nostra [pagina per Educatori](https://developer.mozilla.org/en-US/docs/Learn_web_development/Educators) per ulteriori idee.


## Metti alla prova le tue competenze
Durante il corso troverai diversi articoli progettati per aiutarti a verificare se hai compreso ciò che stiamo insegnando. Esistono due tipi di esercizi:
-	Gli articoli “**Test your skills**” compaiono più frequentemente e verificano la tua conoscenza di una singola funzionalità isolata, come i link HTML, il box model CSS o le funzioni JavaScript.
-	Le “**Challenges**” compaiono meno spesso e verificano la tua capacità di utilizzare più funzionalità insieme per creare un sito web o un programma completo.

La maggior parte delle domande include blocchi di codice HTML/CSS/JavaScript che mostrano il codice di partenza per ogni esercizio. Il modo consigliato per completarli è premere il pulsante “**Play**” in uno dei blocchi per aprire l’esempio nel [MDN Playground](https://developer.mozilla.org/en-US/play), quindi modificare il codice seguendo le istruzioni.
Se commetti un errore, puoi cancellare il tuo lavoro usando il pulsante **Reset** nel Playground. Se rimani bloccato, puoi (di solito) vedere la soluzione in fondo alla sezione della domanda o [chiedere aiuto](https://developer.mozilla.org/en-US/docs/Learn_web_development#contact_us).

***Nota***: *Se preferisci lavorare nel tuo editor o in un editor online (come [CodePen](https://codepen.io/) o [JSFiddle](https://jsfiddle.net/)), puoi copiare il codice dal Playground nel tuo ambiente*.

Alcune domande non includono blocchi di codice iniziali e ti chiedono invece di scaricare file di partenza da usare in locale. A volte ciò è dovuto alla complessità dell’esercizio, altre volte è solo per variare un po’.


## Ottenere i nostri esempi di codice
Gli esempi di codice presenti nella Learning Area sono disponibili su GitHub:
-	Il modo più semplice per ottenerli è [scaricare un file ZIP dell’ultima versione del branch principale](https://codeload.github.com/mdn/learning-area/zip/main).
-	Se conosci Git e GitHub, puoi anche clonare il repository.
